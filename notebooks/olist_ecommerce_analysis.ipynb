```python
import pandas as pd
import numpy as np
```

```python
customer = pd.read_csv('olist_customers_dataset.csv')
geolocation = pd.read_csv('olist_geolocation_dataset.csv')
order_items = pd.read_csv('olist_order_items_dataset.csv')
order_payments = pd.read_csv('olist_order_payments_dataset.csv')
order_reviews = pd.read_csv('olist_order_reviews_dataset.csv')
orders = pd.read_csv('olist_orders_dataset.csv')
products = pd.read_csv('olist_products_dataset.csv')
sellers = pd.read_csv('olist_sellers_dataset.csv')
category_name_translation = pd.read_csv('product_category_name_translation.csv')
```

# SECTION 2: TARGET VARIABLE CONSTRUCTION
________________

```python
orders_delivered = orders[orders['order_status'] == 'delivered'].copy()
```

```python
orders_delivered['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'])
orders_delivered['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'])
orders_delivered['is_late'] = (orders_delivered['order_delivered_customer_date'] > orders_delivered['order_estimated_delivery_date']).astype(int)
```

```python
orders_delivered
```

#
# SECTION 3: JOIN PIPELINE
___

```python
# ORDER ITEMS AGG
# One order can have multiple items → collapse to one row per order
order_items_agg = order_items.groupby('order_id').agg(
    num_items=('order_item_id', 'count'),        
    total_price=('price', 'sum'),                
    total_freight=('freight_value', 'sum')       
).reset_index()

# ORDER PAYMENTS AGG
# One order can have multiple payments (e.g. split payment) → collapse to one row
order_payments_agg = order_payments.groupby('order_id').agg(
    payment_installments=('payment_installments', 'max'),       
    total_payment_value=('payment_value', 'sum'),               
    payment_type=('payment_type', lambda x: x.mode()[0])       
).reset_index()
```

```python
# Rebuild df cleanly
df = orders_delivered.merge(customer, on='customer_id', how='inner')
print(f"Shape after customers join: {df.shape}")

df = df.merge(order_items_agg, on='order_id', how='left')
print(f"Shape after order_items join: {df.shape}")

df = df.merge(order_payments_agg, on='order_id', how='left')
print(f"Shape after payments join: {df.shape}")

# Deduplicated reviews
reviews_clean = order_reviews.sort_values('review_creation_date', ascending=False)\
                              .drop_duplicates(subset='order_id', keep='first')\
                              [['order_id', 'review_score']]

df = df.merge(reviews_clean, on='order_id', how='left')
print(f"Shape after reviews join: {df.shape}")
```

```python
order_product = order_items.groupby('order_id')['product_id'].agg(lambda x:x.mode()[0]).reset_index()
df = df.merge(order_product,on='order_id',how='left')
df = df.merge(products,on='product_id',how='left')
print(f"Shape after products join: {df.shape}")
```

```python
order_items
```

```python
order_seller = order_items.groupby('order_id')['seller_id'].agg(lambda x:x.mode()[0]).reset_index()
df = df.merge(order_seller, on='order_id', how='left')
df = df.merge(sellers, on='seller_id', how='left')
print(f"Shape after sellers join: {df.shape}")
```

```python
df = df.merge(category_name_translation, on='product_category_name', how='left')
print(f"Shape after category translation join: {df.shape}")
```

```python
print(f"Shape after category translation join: {df.shape}")
```

# SECTION 4: DATA CLEANING
___

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
cols_to_drop = ['order_approved_at',
    'order_delivered_carrier_date', 
    'order_delivered_customer_date',
    'payment_installments',
    'total_payment_value',
    'payment_type']
df = df.dropna(subset=cols_to_drop)
df['review_score'] = df['review_score'].fillna(df['review_score'].median())
df['product_category_name'] = df['product_category_name'].fillna('unknown')

for col in ['product_name_lenght', 'product_description_lenght', 
            'product_photos_qty', 'product_weight_g',
            'product_length_cm', 'product_height_cm', 'product_width_cm']:
    df[col] = df[col].fillna(df[col].median())
    
print(f"Remaining nulls: {df.isnull().sum().sum()}")
print(f"Final shape: {df.shape}")    
```

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
df['product_category_name_english'] = df['product_category_name_english'].fillna('unknown')

# Confirm
print(f"Remaining nulls: {df.isnull().sum().sum()}")
print(f"Final shape: {df.shape}")
```

```python
cols_to_drop = ['order_id','customer_id','order_status','product_category_name']
df = df.drop(columns=cols_to_drop)
```

# SECTION 6: MACHINE LEARNING
___

```python
df.dtypes
```

```python
for i in ['order_purchase_timestamp','order_approved_at','order_delivered_carrier_date','order_estimated_delivery_date']:
    df[i] = pd.to_datetime(df[i])
```

```python
df[['order_purchase_timestamp','order_approved_at','order_delivered_carrier_date','order_estimated_delivery_date']]
```

```python
df['approval_time'] = (df['order_approved_at'] - df['order_purchase_timestamp']).dt.total_seconds() / 3600
df['shipping_time'] = (df['order_delivered_carrier_date'] - df['order_approved_at']).dt.total_seconds() / 3600
df['estimated_delivery_days'] = (df['order_estimated_delivery_date'] - df['order_purchase_timestamp']).dt.total_seconds() / 3600 / 24
df['purchase_month'] = pd.to_datetime(df['order_purchase_timestamp']).dt.month
df['purchase_dayofweek'] = pd.to_datetime(df['order_purchase_timestamp']).dt.day_of_week
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder
from sklearn import set_config
set_config(display='text')
```

```python
df.columns
```

```python
cols_to_drop = ['order_purchase_timestamp', 'order_approved_at',
       'order_delivered_carrier_date', 'order_delivered_customer_date',
       'order_estimated_delivery_date','customer_unique_id','customer_city','seller_city',
       'product_id', 'seller_id','order_month']

df_ml = df.drop(columns=cols_to_drop,errors='ignore')
cat_cols = ['customer_state', 'seller_state', 'payment_type', 'product_category_name_english']
le = LabelEncoder()
for col in cat_cols:
    df_ml[col] = le.fit_transform(df_ml[col].astype(str))
    
x = df_ml.drop(columns='is_late')
y = df_ml['is_late']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)

print(f"X_train shape: {x_train.shape}")
print(f"X_test shape: {x_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
rf = RandomForestClassifier(n_estimators=100,random_state=42,class_weight='balanced')
rf.fit(x_train,y_train)
y_pred = rf.predict(x_test)
print(classification_report(y_test, y_pred))
```

```python
feature_importance = pd.DataFrame({'Feature' : x_train.columns ,
                                   'importance' : rf.feature_importances_}).sort_values('importance',ascending=False)
```

```python
fig = px.bar(feature_importance.head(10),
             x='importance',
             y='Feature',
             orientation='h',
             title='Top 10 Feature Importances — Late Delivery Prediction',
             color='importance')
fig.update_layout(yaxis={'categoryorder' : 'total ascending'})
fig.show()
```

```python
x = df_ml.drop(columns=['is_late','review_score'])
y = df_ml['is_late']

X_train, X_test, Y_train, Y_test = train_test_split(x,y,test_size=0.2,random_state=42)
rf2 = RandomForestClassifier(n_estimators=100,random_state=42,class_weight='balanced')
rf2.fit(X_train,Y_train)
Y_pred = rf2.predict(X_test)
print(classification_report(Y_test,Y_pred))
```

```python
feature_importance2 = pd.DataFrame({'feature' : X_train.columns,
                                    'importance' : rf2.feature_importances_}).sort_values('importance',ascending=False)
fig = px.bar(feature_importance2.head(10),
             x='importance',
             y='feature',
             orientation='h',
             title='Top 10 Feature Importances — Late Delivery Prediction (Clean)',
             color='importance')
fig.update_layout(yaxis={'categoryorder' : 'total ascending'})
fig.show()
```
